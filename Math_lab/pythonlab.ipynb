```python
#Plotting using 


import matplotlib.pyplot as plt
import numpy as np

a= np.arange(0, 5, 0.05)
b= np.sin(a*np.pi)

plt.figure(figsize=(3,3))
plt.plot(a,b)
plt.xlabel ("t[s]")
plt.xlabel("s[v]")
plt.title("Sine Wave")
# plt.xlim(0,3)
# plt.ylim(0,2)
plt.grid(True)
plt.show()
plt.axhline(0.8, color= "red")     #horizontal line
plt.axvline( 2, color= "blue") 
plt.savefig("example.png")
```

```python
#Plotting using object_oriented notation

import matplotlib.pyplot as plt
import numpy as np

a= np.arange(0, 5, 0.05)
b= np.sin(a*np.pi)

fig, ax = plt.subplots(
    figsize=(3,2), constrained_layout=True
)
fig.set_facecolor("lightblue")
ax.plot(a,b)
ax.set_xlabel("t[s]")
ax.set_ylabel("s[v]")
ax.set_title("Sine Wave")
```

```python
x= np.linspace(0, 10,100)
y1= np.sin(x)
y2= np.cos(x)

plt.figure(figsize=(5,3))
plt.plot(x,y1, label="sin(x)")
plt.plot(x,y2, label="cos(x)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
x= np.linspace(-10, 10, 200)
y= x**3 -2.0*x**2 -1.0*x + 2.0

plt.figure(figsize= (8,8))

plt.subplot(2,1,1)
plt.title("Default axes limits")
plt.plot(x,y)
plt.grid()

plt.subplot(2,1,2)
plt.title("Customized axes limits")
plt.plot(x,y)
plt.xlim(-5,5)
plt.ylim(-2,3)
plt.grid()
plt.show()
```
