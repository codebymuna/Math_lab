### My solution

```python
# start your code here
x = np.arange(-10, 10, 0.01)

r1 = x < 0.1
r2 = (x >= 0.1) & (x < 4.1)
r3 = (x >= 4.1) & (x < 8.1)
r4 = x >= 8.1

y= r1*0+ r2*(x-0.1)+ r3*(8.1-x)+ r4*0

plt.plot(x, y, label='f(x)', color='Green')

plt.title('Triangle-Shaped Piecewise Function')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.xlim(-10, 10)
plt.grid(True)
plt.legend()
plt.show()
```
